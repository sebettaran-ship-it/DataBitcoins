### Fase 1: Carga de Datos.

```python
#Carga de datos y asignación de tipos.
import pandas as pd
df_data = pd.read_csv(
   "bitcoin_dataset.csv",
   encoding="utf-8",
   dtype={
       "Open": "float64",
       "High": "float64",
       "Low": "float64",
       "Close": "float64",
       "Adj Close": "float64",
       "Volume": "int64"
   },
   parse_dates=["Date"]
)
```

### Fase 2: Limpieza básica.

```python
#Si la fecha fuera texto, la pasamos a formato de fecha.
df_data["Date"] = pd.to_datetime(df_data["Date"])

#Odenamos por fecha.
df_data = df_data.sort_values("Date")

#Seleccionamos las columnas de tipo númericas.
columnas_numericas = df_data.select_dtypes(include="number").columns

#Determinamos nulos y los reemplazamos por el promedio de valores válidos.
for columna in columnas_numericas:
   cantidad_nulos = df_data[columna].isna().sum()
   if cantidad_nulos > 0:
       promedio = df_data[columna].mean() #Cálculo de valor promedio de los valores válidos.
       df_data[columna] = df_data[columna].fillna(promedio)
       #Mansaje de cambios.
       print(
           f"Columna: {columna} | "
           f"Nulos reemplazados: {cantidad_nulos} | "
           f"Promedio utilizado: {promedio:.2f}"
       )

#Eliminamos los duplicados.
df_data = df_data.drop_duplicates()
```

### Fase 3: Preparación para análisis.

```python
#Establecemos la fecha como índice.
df_data = df_data.set_index("Date")

#Se crea un campo de variación diaria.
df_data["Return"] = df_data["Close"].pct_change().fillna(0) #Para evitar que la posición lógica 0 quede con NaN.

#Se crea un campo de diferencia absoluta.
df_data["Price_Diff"] = df_data["Close"] - df_data["Open"]
```
